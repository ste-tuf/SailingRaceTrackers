```python
import pandas as pd
import numpy as np

from datetime import datetime, timedelta

import gpxpy
import pyproj
import json
import xmltodict
import lxml.etree as ET

import os
import shutil

import base64
```

```python
def load_config(path_file):
    with open(path_file, 'r', encoding='latin-1') as f:
            xml_data = f.read()
    report_dict = xmltodict.parse(xml_data)
    return report_dict
```

```python
def load_report(path_file):
    with open(path_file, "r") as f:
        contents = f.read()
    
    # Convert the contents of the text file to a JSON object.
    json_object = json.loads(contents)
    json_object = json_object['reports']
    return json_object
```

```python
report_dict = load_config('../config.json')
report_dict
```

```python
boat_data = report_dict['config']['boats']['boatclass']
boat_data
```

```python
boat_data[0]['boat'][0]['crew']['navigator'][0]
```

```python
boatinfo_json = {}

for i in range(len(boat_data)):
    for j in range(len(boat_data[i]['boat'])):
        boatinfo_json[int(boat_data[i]['boat'][j]['@id'])] = {}
        boatinfo_json[int(boat_data[i]['boat'][j]['@id'])]['boatName'] = boat_data[i]['boat'][j]['@name']
        boatinfo_json[int(boat_data[i]['boat'][j]['@id'])]['skipperNames'] = boat_data[i]['boat'][j]['crew']['navigator'][0]['@fname'] + '_' + \
                                                 boat_data[i]['boat'][j]['crew']['navigator'][0]['@lname'] + '_&_' + boat_data[i]['boat'][j]['crew']['navigator'][1]['@fname'] + '_' + \
                                                 boat_data[i]['boat'][j]['crew']['navigator'][1]['@lname']
        boatinfo_json[int(boat_data[i]['boat'][j]['@id'])]['category'] = report_dict['config']['boats']['boatclass'][i]['@name']
```

```python
with open('boatinfo.json', 'w') as f:
    json.dump(boatinfo_json, f, indent=4, ensure_ascii=False)
```
